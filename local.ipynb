```python
import ollama

response = ollama.chat(model='llama3.2', messages=[
    {
        'role': 'user',
        'content': 'Who is the greatest footballer of all time?',
    }
])

print(response['message']['content'])
```

```python
import ollama
import base64

# Read the image and encode in base64
with open('Cr7.jpg', 'rb') as img_file:
    img_base64 = base64.b64encode(img_file.read()).decode('utf-8')

# Construct the prompt with the image and text
messages = [
    {
        'role': 'user',
        'content': 'Describe this image.',
        'images': [img_base64],  # Note: list of base64-encoded images
    }
]

response = ollama.chat(model='gemma3:4b', messages=messages)

print(response['message']['content'])
```
